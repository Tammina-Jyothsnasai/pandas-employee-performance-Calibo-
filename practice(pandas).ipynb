```python
import pandas as pd

df = pd.read_excel("Employee_Performance_Data.xlsx")
df.head()
```

```python
print("Shape:",df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nDataTypes :")
print(df.dtypes)

print("Information of Dataset:")
df.info()
df.size
```

```python
import numpy as np

print("Summary Statistics:")
display(df.select_dtypes(include=[np.number]).describe())
print("\n Unique Values in Department:")
print(df["Department"].unique())

print("\n Department Counts:")
print(df["Department"].value_counts(dropna=False))
```

```python
print("One column:")
display(df["PerformanceScore"])

print("Multiple Columns:")
display(df[["Name","Department","PerformanceScore"]])
```

```python
print("First Row:")
display(df.loc[0])

print("Specific values:")
print(df.loc[0,"PerformanceScore"])

print("Selected rows and columns:")
display(df.loc[0:4,["Name","PerformanceScore"]])
```

```python
print("First Row:")
display(df.iloc[0])

print("First 5 rows ,first 4 columns:")
display(df.iloc[0:5,0:4])
```

```python
print("Employees with PerformanceScore > 9.5")
display(df[df["PerformanceScore"]>9.5])

print("Employees with Attendance >=98:")
display(df[df["AttendancePct"]>=98])
```

```python
res = df [
    (df["PerformanceScore"]>90) |
    (df["AttendancePct"]>=99)
]
display(res)
print("Sales or HR Employees:")
display(df[df["Department"].isin(["Sales","HR"])])
```

```python
print("Highest Performance scores:")
display(df.sort_values("PerformanceScore",ascending=False).head(10))

print("Lowest Attendence :")
display(df.sort_values("AttendancePct",ascending=True).head(10))
```

```python
work = df.copy()
work["Passed"] = work["PerformanceScore"]>=60
work["Total_Score"]= work["PerformanceScore"]+ work["SatisfactionScore"]
work["Average_Score"] = work[["PerformanceScore","SatisfactionScore"]].mean(axis=1)

display(work[["Name","PerformanceScore","SatisfactionScore","Total_Score","Average_Score","Passed"]].head(10))
```

```python
# Creating a feature with np.where()

work["Performance_level"] = np.where(
    work["PerformanceScore"]>=75,"Excellent",
    np.where(work["PerformanceScore"]>=60,"Good","Needs Improvement")
)
display(work[["Name","PerformanceScore","Performance_level"]].head(10))
```

```python
#Rename Columns
renamed = df.rename(columns = {
    "PerformanceScore":"Final_Performance_Score",
    "TrainingHours": "Daily_Training_Hours"
})
display(renamed.head())
```

```python
#Cleaning of data
# Remove Columns and Rows
# axis = 1/columns -> Remove columns
# axis =0/index -> Remove Rows

example = work.drop(columns=["Total_Score"])
display(example.head())

example = work.drop(index=[0,1])
display(example.head())
```

```python
# Missing Val

print("Missing values by columns:")
print(df.isna().sum())

print("\n Total Missing Values:")
print(df.isna().sum().sum())
```

```python
#Remove rows  with missing values (dropna())

complete_rows = df.dropna()
print("Original shape:",df.shape)
print("After dropna:",complete_rows.shape)
display(complete_rows.head())
```

```python
# Fillna()  we can use mean,median,max,min

filled = df.copy()
filled["AttendancePct"]=filled["AttendancePct"].fillna(
    filled["AttendancePct"].median()
)
filled["TrainingHours"]=filled["TrainingHours"].fillna(
    filled["TrainingHours"].median()
)
filled["Department"]=filled["Department"].fillna("Unknown")

print("Remaining missing Values:")
print(filled.isna().sum())
```

```python
#Forward Fill and Backward Fill

time_df = pd.DataFrame({
    "Day":[1,2,3,4,5],
    "Temperature":[30,np.nan,32,np.nan,35]
})
display(time_df)
print("Forward fill:")
display(time_df.ffill())
print("Backward Fill:")
display("time_df.bfill()")
```
